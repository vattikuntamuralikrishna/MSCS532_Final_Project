```python
import time
import numpy as np

class ParticleAoS:

  def __init__(self, x, y, z, vx, vy, vz, mass):
    self.x = x
    self.y = y
    self.z = z
    self.vx = vx
    self.vy = vy
    self.vz = vz
    self.mass = mass


def benchmark_aos(num_particles, dt, steps):
  # Initializing AoS, creating python object for each particle
  particles = [
      ParticleAoS(
          x=np.random.rand(),
          y=np.random.rand(),
          z=np.random.rand(),
          vx=np.random.rand(),
          vy=np.random.rand(),
          vz=np.random.rand(),
          mass=1.0,
      )
      for _ in range(num_particles)
  ]

  start_time = time.perf_counter() # measuring the update time

  for _ in range(steps): # updating each particle one at a time 
        
    for p in particles:
      p.x += p.vx * dt
      p.y += p.vy * dt
      p.z += p.vz * dt
    
  end_time = time.perf_counter()

  return end_time - start_time


def benchmark_soa_vectorized(num_particles, dt, steps):
    
  # Initializing SoA using contiguous C-arrays
  x = np.random.rand(num_particles)
  y = np.random.rand(num_particles)
  z = np.random.rand(num_particles)
  vx = np.random.rand(num_particles)
  vy = np.random.rand(num_particles)
  vz = np.random.rand(num_particles)

  start_time = time.perf_counter() # measuring the vectorized update time
    
  for _ in range(steps):# updating all particles
    
    x += vx * dt
    y += vy * dt
    z += vz * dt
    
  end_time = time.perf_counter()

  return end_time - start_time


if __name__ == '__main__':
    
  N = 1_000_000  # 1 Million Particles
  
  TIMESTEPS = 10 # running the particles updating for 10 timestamps
  DELTA_T = 0.01

  print(f'=== HPC Data Structure Optimization Benchmark (N={N:,}) ===\n')

  t_aos = benchmark_aos(N, DELTA_T, TIMESTEPS)
  print(f'[Baseline]  AoS Python Object Approach:   {t_aos:.4f} seconds')

  t_soa = benchmark_soa_vectorized(N, DELTA_T, TIMESTEPS)
  print(f'[Optimized] SoA Vectorized Array Approach: {t_soa:.4f} seconds')

  speedup = t_aos / t_soa
  print(f'\n--> Total Speedup Factor: {speedup:.2f}x Performance Improvement')
```
